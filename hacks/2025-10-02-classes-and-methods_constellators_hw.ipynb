### <u>Popcorn Hack 1 - Dice Roller</u>


Instructions:
Finish the code in the cell below (fill in underlines and add lines of code) so it works as described:

- Create a class called Dice that:
    - Has a property for the number of sides.
    - Has a method roll() that returns a random number between 1 and the number of sides.


At the bottom, test your game by making a DiceGame object and calling roll() a few times.

The code is currently NOT functional. When you have completed this popcorn hack, the code should work. Have fun!

```python
%%javascript

class Dice {
  constructor(sides) {
    this.sides = sides;
  }

  roll() {
    return Math.floor(Math.random() * this.sides) + 1;
  }
}

// Test the Dice class:
const myDice = new Dice(6);
console.log("Rolling a 6-sided dice:");

console.log("Roll 1:", myDice.roll());
console.log("Roll 2:", myDice.roll());
console.log("Roll 3:", myDice.roll());

// Test with a different number of sides
const d20 = new Dice(20);
console.log("Rolling a 20-sided dice:", d20.roll());
```

### <u>Popcorn Hack 2 - Pet Simulator</u>


This hack is a bit more complicated than the first. Points will be awarded based on effort and completion.


Analyze the block of code below. Then, complete the following instructions:


1) Create a simple Pet class by filling in the _________.

2) Add at least 2 properties (like hunger, energy, or happiness). EX: Stats decrease over time.

3) Add at least two functioning methods (such as feed, play, or sleep) EX: Buttons increase stats.

4) Give your pet a name and print their current properties.

5) Bonus: Add a "status" method to print all your pet's stats.


The code is currently NOT functional. When you have completed this popcorn hack, the code should work. Have fun!



```python
%%javascript

class Pet {
  constructor(name) {
    this.name = name;
    this.hunger = 100;
    this.energy = 100;
    this.happiness = 100;
  }

  feed() {
    this.hunger = Math.max(0, this.hunger - 20);
    this.energy = Math.min(100, this.energy + 10);
    this.happiness = Math.min(100, this.happiness + 15);
    console.log(`${this.name} has been fed! They are feeling happier and more energetic.`);
  }

  play() {
    this.energy = Math.max(0, this.energy - 25);
    this.hunger = Math.min(100, this.hunger + 15);
    this.happiness = Math.min(100, this.happiness + 30);
    console.log(`You played with ${this.name}! They are very happy but getting tired and hungry.`);
  }

  sleep() {
    this.energy = Math.min(100, this.energy + 40);
    this.hunger = Math.min(100, this.hunger + 10);
    this.happiness = Math.min(100, this.happiness + 5);
    console.log(`${this.name} took a nice nap and feels refreshed!`);
  }

  status() {
    console.log(`=== ${this.name}'s Status ===`);
    console.log(`Hunger: ${this.hunger}/100`);
    console.log(`Energy: ${this.energy}/100`);
    console.log(`Happiness: ${this.happiness}/100`);
    console.log(`========================`);
  }
}

const myPet = new Pet("Andy");
console.log(`Meet your new pet: ${myPet.name}!`);

myPet.status();

myPet.feed();
myPet.play();
myPet.sleep();

console.log("\nAfter some activities:");
myPet.status();
```

```python
%%javascript

class Person {
  constructor(name, age) {
    this.name = name;
    this.age = age;
  }

  greet() {
    console.log(`Hello, my name is ${this.name}!`);
  }
}

const person1 = new Person("Yiming Yin", 15);
console.log(`Name: ${person1.name}, Age: ${person1.age}`);

class Book {
  constructor(title, author, pages) {
    this.title = title;
    this.author = author;
    this.pages = pages;
    this.pagesRead = 0;
  }

  read(pages) {
    this.pagesRead += pages;
    console.log(`You read ${pages} pages of "${this.title}". Total pages read: ${this.pagesRead}`);
  }

  isLongBook() {
    return this.pages > 300;
  }
}

const book1 = new Book("The Great Gatsby", "F. Scott Fitzgerald", 180);
console.log(`Title: ${book1.title}, Author: ${book1.author}, Pages: ${book1.pages}`);
console.log(`Is "${book1.title}" a long book? ${book1.isLongBook()}`);

class Car {
  constructor(make, model, fuel) {
    this.make = make;
    this.model = model;
    this.fuel = fuel;
  }

  drive(distance) {
    const fuelUsed = distance * 0.1;
    this.fuel = Math.max(0, this.fuel - fuelUsed);
    console.log(`Drove ${distance} miles in the ${this.make} ${this.model}. Fuel remaining: ${this.fuel.toFixed(1)} gallons`);
  }

  refuel(amount) {
    this.fuel += amount;
    console.log(`Refueled ${amount} gallons. Current fuel: ${this.fuel} gallons`);
  }

  static info() {
    console.log("Cars are vehicles.");
  }
}

const car1 = new Car("Toyota", "Camry", 15);

class ElectricCar extends Car {
  constructor(make, model, fuel, battery) {
    super(make, model, fuel);
    this.battery = battery;
  }

  charge(amount) {
    this.battery = Math.min(100, this.battery + amount);
    console.log(`Charged ${amount}%. Battery level: ${this.battery}%`);
  }
}

const electricCar1 = new ElectricCar("Tesla", "Model 3", 0, 80);

console.log("\n=== Combined Routine ===");

person1.greet();

car1.drive(50);

electricCar1.charge(15);

book1.read(25);

Car.info();
```
